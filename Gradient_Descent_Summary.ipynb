# Optimization Technique

```python
# (1). Gradient Descent ---> This is the technique to reduce the loss function(Actual - predicted), means (Distance from y_test to y_pred).
```

```python
# We Know that Equation of simple Line :-

# y = mx + b ..................(1)
```

```python
# Loss = (y_act - y_pred)

# So, we will calculate loss with respect to Mean Squared Error :::

# This is the formula for calculating loss w.r.t Means Squared error

# Loss = 1/n * (y_act - y_pred)**2 ..................(2)

# Where n is the number of features

# if n(features) = 1 then whats the loss lets calculate

# loss = 1/1 *(y_act - y_pred)**2

# loss = (y_act - y_pred)**2 ..................................(3)

# Substitute y_pred with simple line equation we can write y_pred = mx + b so

# loss = (y_act - (mx + b))**2

# loss = (y_act -mx -b)**2 .......................................(4)


# x,y is the coordinates of line and m is the weight of line b is the minimum point or constant value that maks the line contribution.


# x,y = Coordinates of lines


# M is the weight


# b is the intercept
```

```python
# Now we will calculate differentiation of loss with respect to Intercept(b)

# dl/db = 2(y_act-mx-b).[dy/db - dmx/db - db/db]

# dl/db = 2(y_act - mx - b).[0 - 0 -1]

# dl/db = -2(y_act - mx - b) ...............................(5)
```

```python
# Now we will calculate differentiation of loss with respect to Weight(m)

# dl/dm = 2(y_act - mx - b).[dy/dm - dmx/dm - db/dm]

# dl/dm = 2(y_act - mx -b).[0 - x -0]

# dl/dm = -2x(y - mx - b) ....................................(6)
```

```python
# b_new = b_old - dl/db in terms of intercept

# Suppose the b_old is 50 and dl/db is 100 then,

# b_new = 50 - 100 = -50 this is called Zig Zag condition,then we add a constant that called learning rate

# Learning_rate = Constant = 0.001(In General)

# b_new = b_old - learning rate * dl/db

# m_new = m_old - dl/dm in terms of weight

# m_new = m_old - learning_rate * dl/dm
```

```python
# If learning_rate >>>> 0 ------> Gradient Exploading
# If learning_rate <<<< 0 ------> Vanishing gradient
```

```python
# So the final equation of new line will be ::::

# y = m_new * x + b_new  ## Thats the final line equation that reduce the loss function.
```

```python
# suppose if we have b_old is 50 and dl/db is 100 on that time we add learning rate to dl/db

b_new = 50 - (0.001)*100

b_new
```
